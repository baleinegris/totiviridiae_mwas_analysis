```python
import polars as pl
```

```python
df = pl.read_parquet("../data/results.parquet")
```

```python
value_mask = df["col_value"].list.eval(pl.element().str.contains("(?i)temp")).list.any()
name_mask = df["col_name"].list.eval(pl.element().str.contains("(?i)temp")).list.any()
temp_rows = df.filter(value_mask | name_mask)
temp_rows
```

```python
# sort by p-value
temp_rows_sorted = temp_rows.sort("pvalue")
temp_rows_sorted.write_parquet("../data/temp_results.parquet")
```

```python
# get rows where bioproject is PRJNA615972 
rows = df.filter(pl.col("bioproject") == "PRJNA615972")
# get rows where col_name is temp_attrib
rows = rows.filter(pl.col("col_name") == ["temp_attrib"])
rows
```

```python
# sort all temp rows by bioproject
rows = temp_rows_sorted.sort("bioproject")
rows.write_parquet("../data/temp_results_sorted.parquet")
```

In order to link multiple MWAS findings together, we plan to use LLMs to generate biologically informed hypotheses to explain differences in *totiviridae* abundance in different metadata groups, using AWS Bedrock. This notebooks is simply a framework to generate one hypothesis, and extrapolates the approximate cost of running this on N rows.

```python
from openai import OpenAI
import os
from openai.providers import bedrock
from dotenv import load_dotenv
load_dotenv()
```

```python
client = OpenAI()

response = client.chat.completions.create(
    model="openai.gpt-6-luna",
    messages=[{"role": "user", "content": "Analyzing SRA metadata and totiviridae abundance, we found the following statistical finding: Samples with the following attributes: genotype_attrib=MATa [HAP1+::NatMX-ACT1pr-Z3EV-ENO2term] [barcode-URA3-Z3EVpr-VMA3] [can1delta::STE2pr-SpHIS5] his3delta1 lyp1delta0 ura3delta0 had a mean concentration of totiviridae of 24.96014784445098 and samples with different attributes had a mean concentration of totiviridae of 6.351905444818192. Please come up with a hypothesis that could explain this finding."}]
)
```

```python
response.choices[0].message.content
```

```python
INPUT = 0.22
OUTPUT = 1.32

completion_tokens = response.usage.completion_tokens
input_tokens = response.usage.prompt_tokens

price = (completion_tokens * OUTPUT + input_tokens * INPUT) / 1000000
price
```

```python
# calculate the price of doing this for all findings rows of data, assuming each row is a separate request
price_total = price * 85326
price_total
```
